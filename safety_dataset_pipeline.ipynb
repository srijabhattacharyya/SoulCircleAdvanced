```python
import pandas as pd

df_safe = pd.read_csv("emotion_dataset.csv")
df_safe = df_safe[["text"]]
df_safe["label"] = 0
```

```python
risk_texts = [
    "I want to kill myself",
    "I don’t want to live anymore",
    "I feel like ending my life",
    "There is no reason to live",
    "I want to disappear forever",
    "I feel completely hopeless",
    "I am tired of everything",
    "I can't go on anymore",
    "Life is not worth living",
    "I feel like giving up completely",
    "I want to hurt myself",
    "I don’t see any future for me",
    "Everything feels pointless",
    "I wish I was dead",
    "I am thinking about suicide"
]

df_risk = pd.DataFrame({
    "text": risk_texts,
    "label": 1
})
```

```python
df_risk = pd.concat([df_risk]*20, ignore_index=True)
```

```python
min_size = min(len(df_safe), len(df_risk))

df_safe = df_safe.sample(min_size)
df_risk = df_risk.sample(min_size)
```

```python
df_final = pd.concat([df_safe, df_risk])
df_final = df_final.sample(frac=1).reset_index(drop=True)
```
