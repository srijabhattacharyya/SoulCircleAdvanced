```python
import subprocess
results = {}
for p in ["python-jose[cryptography]", "passlib[bcrypt]", "python-multipart", "sqlalchemy"]:
    r = subprocess.run(["pip", "install", p], capture_output=True, text=True)
    results[p] = "OK" if r.returncode == 0 else r.stderr[-200:]
for k,v in results.items():
    print(k, "→", v)
```
